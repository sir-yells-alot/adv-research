Task 1 Code

```python
power_data_1 = pd.read_csv("Plant_1_Generation_Data.csv")
weather_data_1 = pd.read_csv("Plant_1_Weather_Sensor_Data.csv")
power_data_2 = pd.read_csv("Plant_2_Generation_Data.csv")
weather_data_2 = pd.read_csv("Plant_2_Weather_Sensor_Data.csv")

# Only this file has a different date time format for some reason
power_data_1['DATE_TIME'] = pd.to_datetime(power_data_1['DATE_TIME'], format="%d-%m-%Y %H:%M")
power_data_2['DATE_TIME'] = pd.to_datetime(power_data_2['DATE_TIME'], format="%Y-%m-%d %H:%M:%S")

weather_data_1['DATE_TIME'] = pd.to_datetime(weather_data_1['DATE_TIME'], format="%Y-%m-%d %H:%M:%S")
weather_data_2['DATE_TIME']= pd.to_datetime(weather_data_2['DATE_TIME'], format="%Y-%m-%d %H:%M:%S")

# Merge power and weather data simply via concatenation
power_data = pd.concat([power_data_1, power_data_2])
weather_data = pd.concat([weather_data_1, weather_data_2])

display(power_data, weather_data)
```

```python

# Filter for specific columns because of duplicates and unneeded columns
power_data_1 = power_data_1[['DATE_TIME', 'DC_POWER', 'SOURCE_KEY', 'PLANT_ID']]
weather_data_1 = weather_data_1[['DATE_TIME', 'IRRADIATION', 'AMBIENT_TEMPERATURE', 'MODULE_TEMPERATURE']]

power_data_2 = power_data_2[['DATE_TIME', 'DC_POWER', 'SOURCE_KEY', 'PLANT_ID']]
weather_data_2 = weather_data_2[['DATE_TIME', 'IRRADIATION', 'AMBIENT_TEMPERATURE', 'MODULE_TEMPERATURE']]

# Merge the power and weather data based on the date and time
merged_data_1 = pd.merge(power_data_1, weather_data_1, on='DATE_TIME', how='inner')
merged_data_2 = pd.merge(power_data_2, weather_data_2, on='DATE_TIME', how='inner')

merged_data = pd.concat([merged_data_1, merged_data_2])

display(merged_data)
```

```python
def show_distributions(df: pd.DataFrame, idx: str):
  # Plot distribution of a specific column
  sns.histplot(df[idx], kde=True)
  plt.title(idx + ' Distribution')
  plt.show()

  # Box plot to detect outliers
  sns.boxplot(x=df[idx])
  plt.title(idx + ' Box Plot')
  plt.show()
```

```python
def show_relationship(df: pd.DataFrame, x: str, y: str):
  sns.scatterplot(data=df, x=x, y=y)
  plt.title(x + ' VS ' + y)
  plt.show()
  print('Correlation ' + df[x].corr(df[y]))
```

```python
merged_data['DC_POWER_NORM'] = merged_data.groupby('SOURCE_KEY')['DC_POWER'].transform(
    lambda x: (x - x.min()) / (x.max() - x.min()) if (x.max() - x.min()) != 0 else 0
) # Normalize DC_POWER as a percent of max power seen by specific inverter

merged_data['HOUR_OF_DAY'] = merged_data['DATE_TIME'].dt.hour # Extract the hour of day as a feature
merged_data['TEMP_DIFF'] = merged_data['MODULE_TEMPERATURE'] - merged_data['AMBIENT_TEMPERATURE'] # How much hotter the module is than the ambient temperature

merged_data_nonzero_power = merged_data[merged_data['DC_POWER'] > 0] # Dataset strictly of datapoints where DC_POWER is nonzero, since I seem to use it a lot
```

```python
merged_data_numerical = merged_data.select_dtypes(include=[np.number])[['DC_POWER', 'IRRADIATION', 'AMBIENT_TEMPERATURE', 'MODULE_TEMPERATURE', 'TEMP_DIFF']] # cant correlate non numerical fields
p_matrix = merged_data_numerical.corr(method=lambda x, y: pearsonr(x, y)[1]) # get p-value matrix
r_matrix = merged_data_numerical.corr() # get r value matrix

mask = np.triu(np.ones_like(r_matrix, dtype=bool))

labels = np.asarray(
    [
        f"r: {r:.2f}\np: {p:.3f}" for r, p in
        zip(r_matrix.values.flatten(), p_matrix.values.flatten())
    ]
  ).reshape(r_matrix.shape)
sns.heatmap(r_matrix, annot=labels, fmt="", cmap="coolwarm", vmin=-1, vmax=1, cbar=True, mask=mask)
plt.title("Correlation Matrix with P-Values")
plt.show()
```

Data cleaning techniques:

```python
from scipy import stats
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
FEATURES = ['IRRADIATION', 'AMBIENT_TEMPERATURE', 'MODULE_TEMPERATURE', 'HOUR_OF_DAY', 'TEMP_DIFF']
```

```python
def split_standard(x, y):
  return train_test_split(x, y, test_size=0.2, random_state=69420)
def split_chronological(x, y):
  return train_test_split(x, y, test_size=0.2, random_state=69420, shuffle=False)
def scale(train_x_df: pd.DataFrame, test_x_df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
  scaler = StandardScaler()
  train_x_scaled = scaler.fit_transform(train_x_df)
  test_x_scaled = scaler.transform(test_x_df)
  return pd.DataFrame(train_x_scaled, columns=train_x_df.columns, index=train_x_df.index), pd.DataFrame(test_x_scaled, columns=test_x_df.columns, index=test_x_df.index)
```

```python
def clean_zscore(df: pd.DataFrame, y_series: pd.Series) -> tuple[pd.DataFrame, pd.Series]:
  z_scores = np.abs(stats.zscore(df))
  threshold = 3.0
  mask = (z_scores < threshold).all(axis=1)
  df_clean = df[mask]
  y_clean = y_series[mask]
  return df_clean, y_clean
```

```python
def modified_zscore(col: pd.Series) -> float:
    median = col.median()
    mad = np.median(np.abs(col - median))

    if mad == 0:
        return np.zeros(len(col))

    return 0.6745 * (col - median) / mad

def clean_modified_zscore(df: pd.DataFrame, y_series: pd.Series) -> tuple[pd.DataFrame, pd.Series]:
  mod_z_scores = np.abs(df.apply(modified_zscore))
  threshold = 3.5
  mask = (mod_z_scores < threshold).all(axis=1)
  df_clean = df[mask]
  y_clean = y_series[mask]
  return df_clean, y_clean
```

```python
def clean_iqr(df: pd.DataFrame, y_series: pd.Series) -> tuple[pd.DataFrame, pd.Series]:
  cols = df

  Q1 = cols.quantile(0.25)
  Q3 = cols.quantile(0.75)
  IQR = Q3 - Q1

  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR

  mask = ((cols >= lower_bound) & (cols <= upper_bound)).all(axis=1)
  df_clean = df[mask]
  y_clean = y_series[mask]
  return df_clean, y_clean
```

```python
def clean_isolation_forest(df: pd.DataFrame, y_series: pd.Series) -> tuple[pd.DataFrame, pd.Series]:
  detector = IsolationForest(contamination="auto", random_state=676767)
  detector.fit(df)

  predictions = detector.predict(df)

  mask = (predictions == 1)
  df_clean = df[mask]
  y_clean = y_series[mask]
  return df_clean, y_clean
```

```python
from sklearn.ensemble import RandomForestRegressor

def fit_random_forest(x: pd.DataFrame, y: pd.Series):
  model = RandomForestRegressor(n_estimators=100, random_state=42069420)
  model.fit(x, y)
  return model
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor
def fit_hist_gradient(x: pd.DataFrame, y: pd.Series):
  model = HistGradientBoostingRegressor(random_state=696969)
  model.fit(x, y)
  return model
```

```python
from sklearn.linear_model import Ridge
def fit_ridge(x: pd.DataFrame, y: pd.Series):
  model = Ridge(alpha=1.0)
  model.fit(x, y)
  return model
```

```python
from sklearn.svm import SVR
def fit_svr(x: pd.DataFrame, y: pd.Series):
  model = SVR(kernel='rbf')
  model.fit(x, y)
  return model
```

```python
#%%script false

from sklearn.metrics import root_mean_squared_error, mean_absolute_error

split_funcs = [split_standard, split_chronological]
clean_funcs = [clean_iqr, clean_zscore, clean_modified_zscore, clean_isolation_forest]
#clean_funcs = [clean_modified_zscore]
fit_funcs = [fit_hist_gradient, fit_random_forest, fit_ridge, fit_svr]
#fit_funcs = [fit_random_forest]

res = []

for split_func in split_funcs:
  data = merged_data.copy()
  x = data[FEATURES]
  y = data['DC_POWER_NORM']

  train_x_initial, test_x_initial, train_y_initial, test_y_initial = split_func(x, y)

  for clean_func in clean_funcs:
    train_x_cleaned, train_y_cleaned = clean_func(train_x_initial.copy(), train_y_initial.copy())

    scaled_train_x, scaled_test_x = scale(train_x_cleaned, test_x_initial)

    for fit_func in fit_funcs:
      model = fit_func(scaled_train_x, train_y_cleaned)

      pred_y = model.predict(scaled_test_x)

      rmse = root_mean_squared_error(test_y_initial, pred_y)
      mae = mean_absolute_error(test_y_initial, pred_y)

      res.append({
        "RMSE": rmse,
        "MAE": mae,
        "Data clean": clean_func.__name__,
        "Split": split_func.__name__,
        "Fit": fit_func.__name__
      })

res = pd.DataFrame(res)
res = res.sort_values(by="RMSE", ascending=True)

res = res.reset_index(drop=True)

display(res)
```
